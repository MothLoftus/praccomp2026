```python
def temp_cov(temp, scale):
    if scale.upper()=="C" or scale.upper()=='CELSIUS':
        print(str(temp)+" C = "+str((9/5)*float(temp)+32)+" F")
    elif scale.upper()=='F' or scale.upper()=='FARENHEIT':
        print(str(temp)+' F = '+str((5/9)*(float(temp)-32))+' C')
    else:
        print("you suck")
```

```python
my_plot_HindfootLxWeight = surveys_df.plot("hindfoot_length", "weight", kind="scatter")
```

```python
my_plot_HindfootLxWeight_bySexes = surveys_df.groupby('sex').plot("hindfoot_length", "weight", kind="scatter")
```

```python
import plotnine as p9
my_plot_HindfootLxWeight_bySexes_p9 = (p9.ggplot(data=surveys_df,
                mapping=p9.aes(x='hindfoot_length', y='weight', color='sex')) +
                p9.geom_point())
my_plot_HindfootLxWeight_bySexes_p9.draw()
```
